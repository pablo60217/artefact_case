```python
import pandas as pd
```

```python
transactions = pd.read_csv("db_transac (1).csv")
```

```python
transactions.head()
```

```python
transactions['is_xb'].value_counts()
```

```python
transactions['is_contactless'].value_counts()
```

```python
transactions.dtypes
```

```python
transactions['prch_date'] = transactions['prch_date'].str.replace('/', '-', regex=False)

transactions['prch_date'].head(20)
```

```python
transactions['prch_time'].head(20)

transactions['prch_time'] = transactions['prch_time'].str.replace('.', ':', regex=False)

transactions['prch_time'].head(20)
```

```python
transactions['product_type'].value_counts()
```

```python
transactions['product_type'] = transactions['product_type'].replace(
    {'Debit': 'Debito', 'Debito': 'Debito', 'Débito': 'Debito',
     'Crédito': 'Credito', 'Credito': 'Credito', 'Credit': 'Credito'}
)

transactions['product_type'].value_counts()
```

```python
transactions['prch_date'] = pd.to_datetime(transactions['prch_date'], errors='coerce')

transactions['prch_datetime'] = transactions['prch_date'] + pd.to_timedelta(transactions['prch_time'], errors='coerce')

transactions[['prch_date', 'prch_time', 'prch_datetime']].head()
```

```python
#years in datetime
transactions['year'] = transactions['prch_datetime'].dt.year
transactions['year'].value_counts()
```

```python
transactions['trimester_year'] = transactions['prch_datetime'].dt.to_period('Q').astype(str)

transactions['trimester_year'].value_counts()
```

```python
#percentiles of amount
transactions['amt'].describe(percentiles=[0.25, 0.5, 0.75, 0.9, 0.95, 0.99])

```

```python
transactions['ctry_card'] = transactions['ctry_card'].replace({'Brazil': 'Brasil', 'Perú': 'Peru', 'Mexico': 'México', 'Panamá':'Panama', 'República Dominicana': 'Republica Dominicana'})

transactions['ctry_card'].value_counts()
```

```python
transactions['ctry_mrch'] = transactions['ctry_mrch'].replace({'Brazil': 'Brasil', 'Perú': 'Peru', 'Mexico': 'México', 'Panamá':'Panama', 'República Dominicana': 'Republica Dominicana'})

transactions['ctry_mrch'].value_counts()
```

```python
#save transactions to csv
transactions.to_csv("transactions_cleaned.csv", index=False)

```
